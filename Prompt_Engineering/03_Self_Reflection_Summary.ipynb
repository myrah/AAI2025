## Source text

A fictional online retailer ran a four-week pilot of an AI assistant for routine delivery questions. The pilot covered 200 tickets. Median first-response time fell from 20 minutes before the pilot to 8 minutes during the pilot. Human agents reviewed every AI draft before sending it. Customer satisfaction was 82% before and 84% during the pilot, but the company did not test whether this difference was statistically significant. The pilot excluded refunds and payment disputes. Staff reported that some drafts contained outdated tracking details, so the team added a required order-status check. The company plans a larger trial before deciding whether to expand the assistant.

## Prompts used

**Initial summary prompt**

Summarize the following pilot for a busy retail manager in two sentences. Focus on the main result and the next step. Source: {source_text}

**Self-critique and revision prompt**

Act as a careful editor. Critique your previous summary against the supplied source, then produce a revised summary. Evaluate: (1) accuracy: preserve the exact metrics and do not imply causation or statistical significance; (2) completeness: include the four-week duration, 200 tickets, human review, excluded refunds/payment disputes, outdated tracking problem and required check, and planned larger trial; (3) clarity and audience: plain language for a nontechnical retail manager; (4) length: 80-110 words, counted by whitespace; (5) format: exactly one paragraph for the revised summary. First list concrete shortcomings of the previous summary by criterion; then provide the revision. Add no external facts. Inputs: source_text={source_text}; previous_summary={previous_summary}.

**Code-generation prompt**

Generate a self-contained Python 3 notebook that preserves the source, initial summary prompt, AI-authored first summary, self-critique prompt, critique, and revised summary. Display both summaries. Count words by whitespace and check the revised summary is one paragraph of 80-110 words. Add a transparent source-fact checklist, and explain that keyword checks do not prove factual accuracy. Store the authored responses as text so the demonstration reruns without a paid API; do not claim a live model call.

```python
source_text = 'A fictional online retailer ran a four-week pilot of an AI assistant for routine delivery questions. The pilot covered 200 tickets. Median first-response time fell from 20 minutes before the pilot to 8 minutes during the pilot. Human agents reviewed every AI draft before sending it. Customer satisfaction was 82% before and 84% during the pilot, but the company did not test whether this difference was statistically significant. The pilot excluded refunds and payment disputes. Staff reported that some drafts contained outdated tracking details, so the team added a required order-status check. The company plans a larger trial before deciding whether to expand the assistant.'
initial_prompt = 'Summarize the following pilot for a busy retail manager in two sentences. Focus on the main result and the next step. Source: {source_text}'
previous_summary = "The retailer's AI pilot cut median first-response time from 20 to 8 minutes, while customer satisfaction rose from 82% to 84%. The company plans a larger trial before deciding whether to expand the assistant."
reflection_template = 'Act as a careful editor. Critique your previous summary against the supplied source, then produce a revised summary. Evaluate: (1) accuracy: preserve the exact metrics and do not imply causation or statistical significance; (2) completeness: include the four-week duration, 200 tickets, human review, excluded refunds/payment disputes, outdated tracking problem and required check, and planned larger trial; (3) clarity and audience: plain language for a nontechnical retail manager; (4) length: 80-110 words, counted by whitespace; (5) format: exactly one paragraph for the revised summary. First list concrete shortcomings of the previous summary by criterion; then provide the revision. Add no external facts. Inputs: source_text={source_text}; previous_summary={previous_summary}.'
print("BEFORE:", previous_summary)
print("Before word count:", len(previous_summary.split()))
```

## Pass the source and previous output to the reflection prompt

```python
reflection_input = reflection_template.format(source_text=source_text, previous_summary=previous_summary)
assert source_text in reflection_input and previous_summary in reflection_input
print(reflection_input)
```

## AI self-critique and revised summary

```python
critique = ['Accuracy: The two reported metrics match the source, but "AI pilot cut" can imply causation. The source reports a before/during change, not a causal evaluation. The satisfaction difference needs the explicit no-significance-test caveat.', 'Completeness: The draft omits four weeks, 200 tickets, human review of every draft, excluded refunds/payment disputes, and outdated tracking details plus the required order-status check. These omissions make the pilot sound broader than it was.', 'Clarity and audience: The draft is readable for a manager. The revision should keep plain language while making the operational limits visible.', 'Length: The draft is below the new 80-110-word requirement. Use the added space for source-backed scope and limitations, not promotional language.', 'Format: Keep the revision as one paragraph and retain the planned larger trial.']
improved_summary = 'In a four-week pilot covering 200 routine delivery tickets, the retailer recorded a drop in median first-response time from 20 to 8 minutes. Human agents reviewed every AI draft before sending it. Customer satisfaction rose from 82% to 84%, but the company did not test whether the difference was statistically significant. Refunds and payment disputes were excluded. Some drafts contained outdated tracking details, prompting a required order-status check. The company plans a larger trial before deciding whether to expand the assistant.'
for item in critique:
    print("-", item)
print("\nAFTER:", improved_summary)
print("After word count:", len(improved_summary.split()))
```

## Validate the requirements
These checks measure format and selected source facts. They are not an automatic truth detector. A manual source comparison additionally checks that the revision avoids causal claims, preserves the uncertainty, and introduces no outside facts.

```python
required_facts = {
    "duration": "four-week", "sample size": "200", "before response time": "20",
    "during response time": "8 minutes", "human review": "reviewed every AI draft",
    "before satisfaction": "82%", "during satisfaction": "84%",
    "uncertainty": "did not test", "excluded topic 1": "Refunds",
    "excluded topic 2": "payment disputes", "known problem": "outdated tracking",
    "mitigation": "order-status check", "next step": "larger trial"
}
for label, phrase in required_facts.items():
    assert phrase.lower() in improved_summary.lower(), label
    print("PASS:", label)
word_count = len(improved_summary.split())
assert 80 <= word_count <= 110
assert len([p for p in improved_summary.split("\n") if p.strip()]) == 1
assert improved_summary != previous_summary
print(f"PASS: {word_count} words (required 80-110); one paragraph; revised text.")
print("13/13 source-fact presence checks passed.")
```

## What improved
The revision adds the pilot's scope, human oversight, excluded topics, known tracking problem, and mitigation. It changes causal-sounding wording to an observed change and restores the statistical-significance caveat. The initial prompt prioritized brevity; the reflection prompt adds measurable requirements and source-grounding. Formatting checks supplement the AI critique; factual accuracy still requires comparison with the source.
